# BAGIAN I --- RECIPE DARI BUKU

## 1. Recipe Audit Station CSV

### Problem

Baca satu file CSV stasiun dan identifikasi masalah pada skema kolom,
rentang waktu, nilai hilang, serta timestamp duplikat. Hasilnya berupa
ringkasan QA/QC, bukan analisis iklim atau prediksi hujan. File sumber
tidak boleh diubah.

### Minimal Theory

Audit data dilakukan sebelum agregasi atau analisis agar kesalahan
struktur, waktu, dan nilai dapat diketahui lebih awal. Timestamp ganda
dapat memengaruhi agregasi, sedangkan nilai hilang atau tipe kolom yang
tidak sesuai dapat mengubah hasil analisis.

### Input dan Constraints

-   `CSV_PATH`: path file CSV.
-   `TIME_COLUMN`: nama kolom waktu.
-   `UNITS`: kontrak satuan untuk kolom yang relevan.
-   Pertahankan file sumber; gunakan pembacaan *read-only*.
-   Jangan menghapus atau memperbaiki baris sumber secara otomatis.
-   Jangan menyimpulkan hasil ilmiah yang tidak diuji oleh audit.

### Prompt dan Procedure

> Buat draft recipe untuk audit QA/QC station CSV. Periksa skema kolom,
> rentang waktu, nilai hilang, timestamp duplikat, dan tipe data tanpa
> mengubah input mentah. Nyatakan input, teori minimum, asumsi,
> prosedur, output yang diharapkan, rencana implementasi Python,
> pemeriksaan kegagalan, serta status eksekusi. Pisahkan spesifikasi
> recipe dari klaim bahwa program telah dijalankan.

Langkah kerja: 1. Baca CSV secara *read-only*. 2. Periksa apakah kolom
wajib tersedia. 3. Validasi tipe data dan rentang timestamp. 4. Hitung
nilai hilang dan timestamp duplikat. 5. Buat laporan QA/QC dan catatan
pemeriksaan. 6. Catat error atau kolom yang tidak sesuai tanpa mengubah
sumber.

### Code / Implementation Plan

Implementasi dapat menggunakan Python dengan `pandas` untuk membaca CSV
dan menghitung ringkasan. Kode disimpan pada path relatif, misalnya
`code/recipes/io/r-io-001.py`. Path ini merupakan rancangan, bukan bukti
bahwa file sudah dibuat.

### Output

-   `results/io/station-summary.md`: laporan Markdown.
-   `results/run-records/r-io-001.json`: run record jika program
    dijalankan.

Laporan mencakup jumlah baris dan kolom, kolom yang ditemukan, rentang
waktu, jumlah nilai hilang, jumlah timestamp duplikat, dan status setiap
pemeriksaan.

### Optional Check / Evidence

Periksa bahwa file input tidak berubah, laporan tidak kosong, jumlah
baris sesuai dengan sumber, dan status eksekusi dicatat. Gunakan
`Not yet executed` jika program belum dijalankan. Label `Executed` hanya
digunakan jika bukti eksekusi tersedia.

-   Jangan menghilangkan fitur atau properti tanpa aturan yang jelas.
-   Jangan menebak CRS atau mengubah urutan koordinat tanpa dasar.
-   Keberhasilan membaca file tidak otomatis membuktikan topologi
    geometri valid.

### Prompt dan Procedure

> inventaris CSV, dan ringkasan JSON. Periksa struktur, tipe geometri,

## 3. Recipe Model Difusi Panas Satu Dimensi

### Problem

Buat recipe untuk mensimulasikan perubahan suhu terhadap posisi dan
waktu berdasarkan profil awal, difusivitas termal, kondisi batas, jarak
grid, dan langkah waktu.

### Minimal Theory

Persamaan difusi panas satu dimensi adalah:

\[
\frac{\partial T}{\partial t}=\alpha\frac{\partial^2T}{\partial x^2}
\]

dengan (T) adalah suhu, (t) waktu, (x) posisi, dan (\alpha)
difusivitas termal.

Untuk skema beda hingga eksplisit standar, parameter kestabilan dapat
ditulis:

\[ r=\frac{\alpha\Delta t}{(\Delta x)^2} \]

Pada kasus satu dimensi dengan skema eksplisit standar, syarat
kestabilan yang umum digunakan adalah (r \le 1/2). Syarat ini
perlu disesuaikan dengan skema numerik yang benar-benar digunakan.

### Input dan Constraints

-   `INITIAL_PROFILE`: profil suhu awal.
-   `ALPHA`: difusivitas termal dan satuannya.
-   `BOUNDARIES`: kondisi batas.
-   `DZ`: jarak grid.
-   `DT`: langkah waktu.
-   Pastikan konsistensi satuan.
-   Jangan mengklaim kalibrasi atau konvergensi tanpa pengujian.

### Prompt dan Procedure

> Buat draft recipe model difusi panas satu dimensi menggunakan profil
> awal `[INITIAL_PROFILE]`, difusivitas `[ALPHA]`, kondisi batas
> `[BOUNDARIES]`, jarak grid `[DZ]`, dan langkah waktu `[DT]`. Sertakan
> persamaan pengatur, konvensi satuan dan tanda, pemeriksaan kestabilan,
> pembanding analitik atau kasus dasar, profil output, grafik, run
> record, serta peringatan jika parameter atau dependency belum
> tersedia. Jangan menyatakan eksekusi tanpa bukti.

Langkah kerja: 1. Validasi parameter dan satuan. 2. Bentuk grid ruang
dan waktu. 3. Terapkan kondisi awal serta kondisi batas. 4. Periksa
kestabilan skema. 5. Jalankan iterasi jika implementasi tersedia. 6.
Bandingkan hasil dengan kasus dasar atau solusi analitik jika
memungkinkan. 7. Simpan profil suhu, grafik, dan catatan eksekusi.

### Code / Implementation Plan

Implementasi dapat menggunakan Python, `numpy` untuk komputasi numerik,
dan `matplotlib` untuk visualisasi. Pilihan metode numerik, parameter,
serta dependency harus dicatat.

### Output

-   Profil suhu terhadap posisi dan waktu.
-   Grafik hasil simulasi.
-   Ringkasan parameter.
-   Run record berisi status dan peringatan.

### Optional Check / Evidence

Periksa kestabilan numerik, kondisi batas, konsistensi satuan, dan
kesesuaian dengan kasus dasar. Jangan mengklaim konvergensi, kalibrasi,
atau keberhasilan eksekusi tanpa bukti yang sesuai.

# BAGIAN II --- RECIPE BARU BUATAN SAYA

## 4. Recipe QC dan Analisis Anomali Gaya Berat Bouguer Lengkap (Lokal 1D/2D Gravity Survey)

### Problem

Membuat workflow reproducible untuk memeriksa kualitas data ukur gaya berat (gravity) lintasan lapang, menghitung koreksi-koreksi standar geofisika (koreksi pasang surut, koreksi drift instrumen, udara bebas, dan Bouguer), memvisualisasikan profil Anomali Bouguer Lengkap (ABL), serta memperkirakan kontras densitas struktur bawah permukaan.

Masalah yang diperiksa meliputi nilai pembacaan alat yang hilang, nilai drift non-linear, kesalahan koordinat/elevasi stasiun, elevasi bernilai negatif yang tidak sesuai, serta lonjakan anomali akibat ketidaktepatan koreksi medan (terrain correction). Tujuannya adalah menghasilkan profil anomali yang bersih dan valid sebelum dilanjutkan ke tahap pemodelan maju 2D/3D atau pemisahan anomali regional-sisa.

### Minimal Theory

Pembacaan gravitometer terukur ($g_{obs}$) dikoreksi terhadap drift alat dan pasang surut gravitasi Bulan-Matahari ($g_{corr}$). Anomali Bouguer Sederhana ($CBA$) dihitung menggunakan rumus:

\[ CBA = g_{obs} - g_n + FAC - BC \]

dengan $g_n$ adalah gaya berat teoritis (formulasi IGF1967/WGS84), $FAC = 0.3086 \cdot h$ adalah Koreksi Udara Bebas (mGal), $BC = 0.04193 \cdot \rho \cdot h$ adalah Koreksi Bouguer Slab (mGal) dengan densitas acuan batuan kerak (umumnya $\rho = 2.67 \text{ g/cm}^3$), dan $h$ adalah elevasi stasiun (m).

Untuk Anomali Bouguer Lengkap ($CBA_{complete}$), ditambahkan Koreksi Medan ($TC$):

\[ ABL = CBA + TC \]

Anomali gaya berat mencerminkan variasi lateral dari kontras densitas ($\Delta\rho$) bawah permukaan. Nilai anomali tunggal tidak secara pasti menentukan jenis batuan karena adanya ambiguitas pemodelan potensiil (*non-uniqueness*).

### Input dan Constraints

Input utama:
- `DATA_PATH`: lokasi file CSV data pengukuran gaya berat.
- `STATION_ID`: nama atau nomor stasiun pengukuran.
- `LATITUDE`, `LONGITUDE`: koordinat geospasial stasiun (derajat desimal).
- `ELEVATION`: elevasi stasiun di atas permukaan laut (m).
- `G_OBS`: pembacaan nilai gaya berat terukur (mGal).
- `TIME`: waktu pembacaan untuk perhitungan drift dan pasang surut.
- `DENSITY`: densitas rata-rata kerak acuan (misal: 2.67 g/cm³).

Input tambahan jika tersedia:
- `TERRAIN_CORR`: nilai koreksi medan dari DEM/Hammer chart.
- `BASE_STATION`: data pembacaan stasiun base awal dan akhir.

Batasan:
1. Berkas data mentah bersifat *read-only*.
2. Drift instrumen dihitung hanya jika terdapat pembacaan berulang pada stasiun base.
3. Koordinat dan elevasi harus tervalidasi dalam rentang wilayah eksplorasi yang rasional.
4. Nilai $ABL$ digambarkan sebagai fungsi dari jarak lintasan atau posisi koordinat.
5. Dilarang mengklaim identifikasi struktur patahan atau batas batuan tanpa analisis turunan (derivatif) dan konfirmasi geologi.

### Prompt dan Procedure

**Prompt yang dapat digunakan kembali:**

> Anda bertindak sebagai asisten pengolahan data geofisika metode gaya berat (gravitasi). Baca berkas `[DATA_PATH]` secara read-only. Periksa kelengkapan kolom wajib (`STATION_ID`, `LATITUDE`, `ELEVATION`, `G_OBS`, `TIME`), validasi tipe data, deteksi nilai hilang, dan identifikasi pencilan (*outliers*) pada elevasi maupun pembacaan alat.
>
> Buat laporan QA/QC data mentah. Lakukan kalkulasi koreksi drift, gaya berat normal ($g_n$), koreksi udara bebas ($FAC$), koreksi Bouguer ($BC$), serta tambahkan koreksi medan ($TC$) jika tersedia. Tampilkan grafik profil Anomali Bouguer Lengkap ($ABL$) terhadap jarak lintasan.
>
> Pisahkan secara jelas antara data hasil observasi, nilai hasil koreksi numerik, dan interpretasi struktur bawah permukaan. Simpan laporan dan grafik pada direktori terdefinisi, serta buat run record jika program dijalankan.

**Procedure:**
1. Baca data mentah gaya berat secara *read-only*.
2. Lakukan audit skema: periksa kolom wajib, tipe data, dan pencilan pada koordinat/elevasi.
3. Hitung linieritas drift berdasarkan stasiun base awal dan akhir lintasan.
4. Hitung nilai gaya berat teoritis ($g_n$), $FAC$, dan $BC$ dengan $\rho = 2.67 \text{ g/cm}^3$.
5. Jumlahkan seluruh koreksi untuk menghasilkan nilai Anomali Bouguer Lengkap ($ABL$).
6. Plot profil $ABL$ beserta lintasan spasial stasiun.
7. Evaluasi variasi nilai anomali dan hitung rentang residual/pencilan data.
8. Susun laporan QC, tabel hasil perhitungan anomali, serta simpan metadata eksekusi.

### Code / Implementation Plan

Rencana implementasi menggunakan Python:
- `pandas`: manipulasi data tabular dan pembuatan laporan QC.
- `numpy`: kalkulasi matematis rumus gaya berat teoritis dan koreksi.
- `matplotlib` / `seaborn`: plot profil $ABL$ dan peta lintasan stasiun.
- `json` & `pathlib`: menyimpan catatan eksekusi dan metadata.

Rancangan struktur direktori proyek:

``` text
gravity-survey-qc/
├── data/
│   └── gravity-survey-data.csv
├── code/
│   └── gravity_qc_processing.py
└── results/
    ├── gravity-qc-report.md
    ├── complete-bouguer-profile.png
    ├── gravity-reduction-results.csv
    └── run-record.json
```

Struktur di atas merupakan rancangan acuan yang akan diverifikasi saat eksekusi.

### Output

1. `gravity-qc-report.md`: laporan QA/QC berisi statistik data mentah, jumlah stasiun valid, serta evaluasi nilai drift alat.
2. `complete-bouguer-profile.png`: grafik profil Anomali Bouguer Lengkap ($ABL$) vs Jarak Stasiun (m).
3. `gravity-reduction-results.csv`: tabel lengkap memuat stasiun, elevasi, $g_{obs}$, $g_n$, $FAC$, $BC$, $TC$, dan $ABL$.
4. `run-record.json`: catatan eksekusi memuat waktu eksekusi, parameter densitas yang digunakan, serta log sistem.

### Optional Check / Evidence

Checklist pemeriksaan:
- [ ] File CSV input tidak diubah (*read-only*).
- [ ] Formula gaya berat normal ($g_n$) tervalidasi sesuai standar IGF.
- [ ] Nilai densitas acuan terdefinisi eksplisit.
- [ ] Grafik disajikan lengkap dengan label sumbu, judul, dan legenda.
- [ ] Ambiguitas pemodelan metode potensiil dicantumkan dalam catatan.
- [ ] Status eksekusi ditandai secara tepat (`Draft` / `Not yet executed` / `Executed`).

Status eksekusi yang diperbolehkan:
- `Draft`: rancangan spesifikasi recipe sedang disusun.
- `Not yet executed`: skrip program belum dijalankan.
- `Partial`: kalkulasi reduksi selesai, namun koreksi medan belum diterapkan.
- `Executed`: skrip telah berhasil dijalankan dan seluruh output tersedia.

### Keterbatasan dan Risiko

Metode Gaya Berat memiliki sifat *non-uniqueness* (ambiguitas tinggi) di mana berbagai kombinasi bentuk, kedalaman, dan kontras densitas struktur bawah permukaan dapat menghasilkan kurva respon anomali yang identik di permukaan. Ketidaktepatan penentuan nilai densitas rerata Bouguer maupun koreksi topografi lokal ($TC$) dapat memunculkan anomali semu yang menyerupai struktur geologi asli. Oleh karena itu, interpretasi anomali Bouguer wajib dibatasi sebagai analisis kualitatif indikatif sebelum dilakukan pemodelan kuantitatif terintegrasi.

### Ringkasan Recipe Baru

Recipe ini menyusun kerangka kerja baku dalam tahapan koreksi dan pemrosesan data gaya berat eksplorasi. Dimulai dari pengujian kualitas pembacaan instrumen, penerapan koreksi numerik bertahap, pembuatan profil anomali yang bebas dari efek luar, hingga pendokumentasian bukti pengolahan secara transparan dan terstruktur.
