# Definisi Struktur Prompt

**ROLE (R)**

Role adalah bagian prompt yang menentukan **peran atau sudut pandang yang
harus digunakan AI** dalam mengerjakan suatu tugas. Role membantu AI
menyesuaikan istilah, fokus, prioritas, dan cara memberikan respons sesuai
dengan bidang yang diminta.

---

**TASK (T)**

Task adalah bagian prompt yang menjelaskan **tugas utama yang harus
dikerjakan oleh AI**. Task harus ditulis secara jelas dan spesifik agar
pekerjaan yang diminta dapat dipahami dan diperiksa hasilnya.

---

**CONTEXT (C)**

Context adalah bagian prompt yang memberikan **latar belakang dan tujuan
dari tugas**. Context membantu AI memahami situasi, bidang kajian, tahapan
pekerjaan, serta tujuan ilmiah dari tugas yang diberikan.

---

**INPUT (I)**

Input adalah bagian prompt yang menjelaskan **data, informasi, variabel,
sumber, atau bahan yang digunakan AI dalam mengerjakan tugas**. Input perlu
disertai informasi yang relevan seperti satuan, koordinat, waktu, dan sumber
data jika diperlukan.

---

**CONSTRAINTS (K)**

Constraints adalah bagian prompt yang berisi **batasan, aturan, metode, atau
ketentuan yang harus dipatuhi AI** selama mengerjakan tugas. Batasan dapat
mencakup metode yang digunakan, software, library, satuan, konvensi ilmiah,
serta larangan mengubah data mentah.

---

**OUTPUT (O)**

Output adalah bagian prompt yang menentukan **bentuk dan isi hasil yang
diharapkan dari AI**. Output dapat berupa tabel, grafik, perhitungan,
penjelasan, interpretasi, kode, atau bentuk hasil lainnya sesuai kebutuhan.

---

**QUALITY CRITERIA (Q)**

Quality Criteria adalah bagian prompt yang menentukan **kriteria kualitas
yang harus dipenuhi oleh hasil AI**. Kriteria ini dapat mencakup akurasi,
konsistensi satuan dan notasi, kejelasan asumsi, alur penalaran, serta
keterlacakan hasil terhadap data atau sumber yang digunakan.

---

**VERIFICATION (V)**

Verification adalah bagian prompt yang digunakan untuk **memeriksa kembali
hasil yang telah diperoleh AI**. Verifikasi dapat mencakup pengecekan
perhitungan, satuan, konsistensi hasil, kesesuaian dengan teori, atau
perbandingan dengan nilai referensi.

# Prompt 1

Bertindak sebagai **[ROLE GEOFISIKA]**, misalnya:
 "Geolistrik Resistivitas"

---

**TASK (T)**

[VERB YANG DAPAT DIPERIKSA] + [OBJECT] + [UNIT/DOMAIN] + [KONDISI PENERIMAAN]

Contoh pola verba geofisika:
- menghitung resistivitas berdasarkan data arus dan beda potensial,
- mengidentifikasi perubahan nilai resistivitas,
- menentukan batas lapisan,
- membuat model bawah permukaan,
- menginterpretasikan kemungkinan material penyusun lapisan.

Misal:

> "Analisis data geolistrik resistivitas untuk menentukan variasi nilai resistivitas terhadap kedalaman, mengidentifikasi batas antar lapisan bawah permukaan, dan memperkirakan karakteristik material penyusun setiap lapisan."

Jika task memiliki beberapa hasil, pecah menjadi langkah/recipe terpisah.

---

**CONTEXT (C)**

Jelaskan tujuan ilmiah, tahap workflow, dan audiens:

- **Topik:** Metode Geolistrik Resistivitas
- **Tahap:** Pengolahan dan interpretasi data bawah permukaan
- **Prasyarat yang sudah dikuasai:** konsep arus listrik, beda potensial, resistivitas, faktor geometri, dan dasar interpretasi geolistrik.
- **Tujuan:** mengetahui variasi sifat kelistrikan material bawah permukaan.
- **Batas relevansi:** hanya gunakan informasi yang berkaitan dengan pengolahan dan interpretasi data resistivitas.
---

**INPUT (I)**

- **Source utama:** data hasil survei geolistrik.
- **Variable/field penting & unit:** arus listrik (I), beda potensial (ΔV), faktor geometri (K), jarak elektroda, koordinat titik pengukuran, dan nilai resistivitas.
- **Coordinate/time convention:** gunakan sistem koordinat yang tercantum pada data.
- **Provenance:** sumber data dan lokasi pengukuran harus dicantumkan apabila tersedia.
- **Input yang TIDAK boleh dipakai:** data tambahan yang tidak diberikan pengguna.

---

**CONSTRAINTS (K)**

- **Method yang diwajibkan:**Gunakan persamaan dasar metode resistivitas dan Pertahankan satuan Ωm untuk nilai resistivitas.
- **Method yang dilarang:** Jangan mengubah data mentah dan Jangan menentukan jenis batuan secara mutlak hanya berdasarkan nilai resistivitas.
- **Batas software/library:** Microsoft Excel atau Python dengan NumPy,
  pandas, dan matplotlib.
- **Larangan side effect:** Jika terdapat data yang tidak lengkap, nyatakan data yang diperlukan daripada membuat nilai sendiri.

---

**OUTPUT (O)**

Bentuk deliverable yang diminta:

1. Tabel hasil perhitungan resistivitas.
2. Grafik hubungan resistivitas dengan kedalaman.
3. Identifikasi lapisan bawah permukaan.
4. Perkiraan ketebalan setiap lapisan.
5. Interpretasi kemungkinan material penyusun.
6. Kesimpulan hasil analisis.

---

**QUALITY CRITERIA (Q)**

- Perhitungan sesuai dengan persamaan geolistrik.
- Satuan konsisten.
- Data mentah tetap dipertahankan.
- Setiap asumsi dijelaskan.
- Alur analisis jelas: data → perhitungan → model → interpretasi.
- Interpretasi dibedakan dari hasil perhitungan.

---

**VERIFICATION (V)**

Jika menyangkut rumus atau nilai kuantitatif:

- Minta AI menunjukkan pengecekan perhitungan dan satuan.
- Periksa kembali perhitungan resistivitas.
- Periksa konsistensi satuan.
- Pastikan nilai resistivitas sesuai dengan data arus dan beda potensial.
- Periksa kembali batas lapisan yang diperoleh.
- Pastikan interpretasi tidak melebihi informasi yang diberikan oleh data.

# Prompt 2

Bertindak sebagai **[ROLE GEOFISIKA]**, misalnya:
- "ANALISIS SEISMIK REFRAKSI"

---

**TASK (T)**

Analisis data seismik refraksi untuk menentukan kecepatan rambat gelombang P pada setiap lapisan dan memperkirakan kedalaman batas antar lapisan.

Contoh:
- mengidentifikasi waktu tiba gelombang,
- membuat grafik travel time terhadap jarak,
- menentukan kemiringan kurva,
- menghitung kecepatan gelombang,
- memperkirakan kedalaman lapisan,
- membuat model bawah permukaan.

---

**CONTEXT (C)**

**Topik:** Seismik Refraksi  
**Tahap:** Pengolahan data waktu tiba gelombang  
**Prasyarat:** konsep gelombang P, waktu tempuh, kecepatan gelombang, dan hukum Snell.
**Batas relevansi:** hanya membahas data seismik refraksi yang diberikan.

---

**INPUT (I)**

**Source utama:** Data waktu tiba gelombang P dari beberapa geofon.
**Variable/field penting & unit:**
- jarak sumber ke geofon, posisi geofon, waktu tiba gelombang P, dan jumlah geofon.
**Coordinate/time convention:** gunakan sistem koordinat atau posisi lintasan yang diberikan. 
**Provenance:** sumber data, lokasi survei, dan konfigurasi akuisisi jika tersedia.  
**Input yang TIDAK boleh dipakai:** data seismik tambahan yang tidak diberikan.

---

**CONSTRAINTS (K)**

**Method yang diwajibkan:** Gunakan prinsip dasar metode seismik refraksi dan Gunakan satuan yang konsisten.
**Method yang dilarang:**Jangan mengubah waktu tiba pada data mentah dan Jangan menentukan jenis batuan secara pasti hanya berdasarkan kecepatan gelombang.
**Batas software/library:** Excel, Python
**Ambiguity rule:** Jika terdapat data yang tidak mencukupi, jelaskan keterbatasannya 


---

**OUTPUT (O)**

1. Tabel jarak dan waktu tiba.
2. Grafik travel time terhadap jarak.
3. Kecepatan gelombang setiap lapisan.
4. Perkiraan kedalaman batas lapisan.
5. Model lapisan bawah permukaan.
6. Interpretasi hasil.
7. Kesimpulan.
---

**QUALITY CRITERIA (Q)**

- Kecepatan diperoleh dari hubungan jarak dan waktu yang benar.
- Satuan konsisten.
- Grafik memiliki sumbu dan satuan yang jelas.
- Model lapisan sesuai dengan hasil perhitungan.
- Asumsi dijelaskan.
- Interpretasi tidak melebihi informasi yang tersedia.

---

**VERIFICATION (V)**

- Periksa kembali gradien grafik travel time.
- Periksa satuan kecepatan.
- Verifikasi kedalaman lapisan menggunakan persamaan yang digunakan.
- Pastikan model bawah permukaan konsisten dengan data waktu tiba.
- Bandingkan hasil dengan konsep dasar seismik refraksi.

# Prompt 3

Bertindak sebagai **[ROLE GEOFISIKA]**, Bertindaklah sebagai analis data geomagnetik yang memiliki kemampuan dalam pengolahan medan magnet dan interpretasi anomali magnetik untuk kajian bawah permukaan. misalnya:
- "INTERPRETASI DATA GEOMAGNETIK"

---

**TASK (T)**

Analisis data geomagnetik untuk memperoleh nilai anomali magnetik, mengidentifikasi pola anomali, dan menentukan kemungkinan adanya struktur geologi bawah permukaan.

Contoh:
- memeriksa data medan magnet,
- melakukan koreksi yang diperlukan,
- menghitung nilai anomali,
- membuat grafik atau peta anomali,
- mengidentifikasi pola anomali,
- melakukan interpretasi geologi.

---

**CONTEXT (C)**

**Topik:** Metode Geomagnetik  
**Tahap:** Pengolahan dan interpretasi data medan magnet.  
**Prasyarat:** medan magnet bumi, medan magnet total, anomali magnetik, dan dasar interpretasi geofisika.
**Batas relevansi:** hanya gunakan informasi yang berkaitan dengan data geomagnetik.

---

**INPUT (I)**

**Source utama:** data hasil pengukuran medan magnet.
**Variable/field penting & unit:**
- koordinat, medan magnet total, waktu pengukuran, dan data base station apabila tersedia.
**Coordinate/time convention:**  gunakan sistem koordinat yang tersedia.
**Provenance:**  lokasi, tanggal pengukuran, dan sumber data.
**Input yang TIDAK boleh dipakai:** data magnetik tambahan yang tidak diberikan.

---

**CONSTRAINTS (K)**

**Method yang diwajibkan:** Gunakan prinsip dasar metode geomagnetik.
**Method yang dilarang:** Jangan mengubah data mentah dan Jangan menyimpulkan jenis batuan secara mutlak hanya berdasarkan anomali magnetik.
**Batas software/library:** Jika data tidak lengkap, nyatakan keterbatasannya.
**Ambiguity rule:** Jika koreksi diperlukan, jelaskan jenis koreksi dan dasar penggunaannya.


---

**OUTPUT (O)**

1. Tabel data medan magnet.
2. Data hasil koreksi.
3. Nilai anomali magnetik.
4. Grafik atau peta anomali.
5. Identifikasi pola anomali.
6. Interpretasi kemungkinan struktur geologi.
7. Kesimpulan.
---

**QUALITY CRITERIA (Q)**

- Perhitungan anomali sesuai metode yang digunakan.
- Satuan konsisten.
- Posisi titik pengukuran tidak berubah.
- Peta/grafik memiliki legenda, skala, dan satuan yang jelas.
- Interpretasi berdasarkan pola anomali dan data pendukung.
- Hasil perhitungan dipisahkan dari interpretasi.

---

**VERIFICATION (V)**

- Periksa nilai medan magnet sebelum dan sesudah koreksi.
- Periksa satuan setiap parameter.
- Pastikan perhitungan anomali benar.
- Periksa kembali pola pada grafik atau peta.
- Pastikan interpretasi sesuai dengan prinsip geomagnetik.
